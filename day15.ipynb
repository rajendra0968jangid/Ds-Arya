```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'student': [
        'Rahul', 'Priya', 'Amit', 'Neha', 'Rohit',
        'Sneha', 'Karan', 'Pooja', 'Vikas', 'Anjali'
    ],
    'gender': [
        'Male', 'Female', 'Male', 'Female', 'Male',
        'Female', 'Male', 'Female', 'Male', 'Female'
    ],
    'study_hours': [2, 5, 3, 6, 1, 4, 2, 5, 3, 4],
    'attendance': [70, 90, 75, 95, 60, 85, 65, 92, 80, 88],
    'math': [55, 85, 65, 92, 40, 78, 50, 88, 70, 82],
    'science': [60, 88, 70, 90, 45, 80, 55, 85, 72, 86],
    'english': [65, 92, 68, 94, 50, 82, 48, 90, 75, 84]
}

df = pd.DataFrame(data)

print(df)
```

```python
df['total_marks'] = df['math'] + df['english'] + df['science']
df['avg_marks'] = df[['math','english','science']].mean(axis=1)

print(df[['student','total_marks','avg_marks']])
```

```python
# top performing students
print(df.sort_values('total_marks',ascending=False).head(3))
```

```python
# Find students who scored above 75 on average:-
print(df[df['avg_marks'] > 75])
```

```python
# calculate avg marks by gender
print(df.groupby('gender')['avg_marks'].mean())
```

```python
#Check for missing values :
print(df.isnull().sum())
```

```python
#Check for duplicates values :
print(df.duplicated().sum())
```

```python
# numpy for numerical calcal:
marks = df['avg_marks']
print("Average marks:", np.mean(marks))
print("Highest marks:", np.max(marks))
print("Lowest marks:", np.min(marks))
print("Standard deviation:", np.std(marks))
```

```python
#Count students scoring at least 75 on average:
count = np.sum(marks >= 75)

print("Students scoring 75 or above:", count)
```

```python
#Visualize the results
plt.figure(figsize=(10, 5))

plt.bar(df['student'], df['avg_marks'])

plt.title('Average Marks of Students')
plt.xlabel('Student')
plt.ylabel('Average Marks')
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
#Matplotlib histogram
plt.figure(figsize=(8, 5))

plt.hist(df['avg_marks'], bins=5, edgecolor='black')

plt.title('Distribution of Average Marks')
plt.xlabel('Average Marks')
plt.ylabel('Number of Students')

plt.show()
```

```python
#Seaborn scatter plot
sns.scatterplot(
    data=df,
    x='study_hours',
    y='avg_marks'
)

plt.title("Study Hours vs Average Marks")
plt.show()
```

```python
#Seaborn box plot
sns.boxplot(
    data=df,
    x='gender',
    y='avg_marks'
)

plt.title('Average Marks by Gender')
plt.show()
```
