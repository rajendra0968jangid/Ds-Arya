Relational plots

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips = sns.load_dataset("tips")
tips.rename(columns={"sex":"gender"},inplace=True)
tips
```

```python
# scatter plot -> axis level function
sns.scatterplot(data=tips,x="total_bill",y="tip",hue="gender",style="time",size="size")
```

```python
# relplot -> figure level function -> squre shape

sns.relplot(data=tips,x="total_bill",y="tip",kind="scatter",hue="gender",style="time",size="size")
```

```python
# line plot
gap = px.data.gapminder()
gap
```

```python
#line plot
tmp_df = gap[gap['country'] == 'India']
```

```python
tmp_df
```

```python
# axis level function
sns.lineplot(data=tmp_df,x="year",y="lifeExp")
```

```python
# using relplot
sns.relplot(data=tmp_df,x='year',y='lifeExp',kind="line")
```

```python
# hue -style
tmp_df = gap[gap['country'].isin(['India','Pakistan','China'])]
tmp_df
```

```python
sns.relplot(kind="line",data=tmp_df,x="year",y="lifeExp",hue="country")
```

```python
sns.lineplot(data=tmp_df,x="year",y="lifeExp",hue="country")
```

```python
# hue -style
tmp_df = gap[gap['country'].isin(['India','Germany','Brazil'])]
tmp_df
```

```python
sns.relplot(kind="line",data=tmp_df,x="year",y="lifeExp",hue="country",style="continent")
```

```python
sns.relplot(kind="line",data=tmp_df,x="year",y="lifeExp",hue="country",style="continent",size="continent")
```
